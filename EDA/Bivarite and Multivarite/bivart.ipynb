```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
train = pd.read_csv('train.csv')
```

```python
tips = sns.load_dataset('tips')
tips
```

```python
iris = sns.load_dataset('iris')
iris
```

```python
flights = sns.load_dataset('flights')
flights
```

# 1 .Scatterplot(Numerical - Numerical)

```python
sns.scatterplot(x='total_bill', y='tip', data=tips, hue='sex', style='smoker', size='size')
```

# Barplot (Numerical - Categorical)

```python
train.head()
```

```python
sns.barplot(x='Pclass', y='Age', hue='Sex', data=train)
plt.title("Average Age by Pclass and Sex")
plt.show()
```

3. Box PLot (Numerical - Categorical)

```python
sns.boxplot(x='Pclass', y='Age', hue='Survived', data=train)
plt.title("Age Distribution by Pclass and Survival")
plt.show()
```

4. Distplot (Numerical - Categorical)

```python
sns.kdeplot(train[train['Survived'] == 0]['Age'].dropna(), label='Not Survived')
sns.kdeplot(train[train['Survived'] == 1]['Age'].dropna(), label='Survived')

plt.xlabel("Age")
plt.ylabel("Probability Density")
plt.title("Age PDF by Survival")
plt.legend()
plt.show()
```

5. HeatMap (Categorical - Categorical)

```python
sns.heatmap(pd.crosstab(train['Pclass'], train['Survived']), annot=True, fmt='d')
```

```python
train.groupby('Pclass')['Survived'].mean() * 100
```

6.Cluster (Categorical - Categorical)

```python
sns.clustermap(pd.crosstab(train['SibSp'],train['Survived']))
```

7. Pairplot

```python
sns.pairplot(iris,hue='species')
```

8. Lineplot (Numerical - Numerical)

```python
flights.head
```

```python
new = flights.groupby('year')['passengers'].sum().reset_index()
```

```python
sns.lineplot(x='year', y='passengers', data=new)
plt.title("Total Passengers per Year")
plt.xlabel("Year")
plt.ylabel("Passengers")
plt.show()
```
