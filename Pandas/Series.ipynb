Pandas Series

A pandas series is like a column in a table . Its a 1 D array holding data of any type.

```python
import numpy as np
```

```python
import pandas as pd
```

```python
#series
coountry = pd.Series(['India','USA','UK','Russia','Srilanka'])
pd.Series(coountry)
```

```python
# interger

runs = pd.Series([100,200,300,400,500])
pd.Series(runs)
```

```python
#custom index
marks = [90,85,78,88,92]
subject=['Maths','Physics','Chemistry','Biology','English']

pd.Series(marks, index=subject)
```

```python
# setting a name
pd.Series(marks, index=subject, name="Jayesh ke mark")
```

Series from dict

```python
Marks = {
    "maths":58,
    "english":45,
    "science":56,
    "hindi":56
}

marks_series = pd.Series(Marks, name ="jayesh ke marks")
marks_series
```
