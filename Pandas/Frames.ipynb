Filtering a DataFroame

```python
ipl.head(2)
```

```python
# find all the final winners
mask = ipl ['MatchNumber'] == 'Final'
new_df = ipl[mask]
new_df[["Season", "WinningTeam"]]

ipl[ipl['MatchNumber'] == 'Final'][["Season", "WinningTeam"]]
```

```python
# how many super over fineshes have occured in IPL history
ipl[ipl['SuperOver'] == 'Y'].shape[0]
ipl[ipl['SuperOver'] == 'Y']
```

```python
# how many matches had csk won in kolkata
ipl[(ipl['City'] == 'Kolkata') & (ipl['WinningTeam'] == 'Chennai Super kings')].shape[0]
```

```python
# toss winner is match winner in percentage
(ipl[ipl['TossWinner'] == ipl['WinningTeam']].shape[0]/ipl.shape[0])*100
```

```python
# movies with rating higher than 8 and votess>10000
movies[(movies['imdb_rating'] > 8.2) & (movies['imdb_votes'] > 10000)].shape[0]
```

```python
# action movies with rating higher than 7.5
mask1 = movies['genres'].str.split('|').apply(lambda x: 'Romance' in x)
mask2 = movies['imdb_rating'] > 7.5
movies[mask1 & mask2]
```
